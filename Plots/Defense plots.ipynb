### 0 Imports and settings

```python
# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import seaborn as sns
from typing import Literal

import matplotlib as mpl
from matplotlib.axes._axes import Axes
from matplotlib.figure import Figure
import matplotlib.patches
from matplotlib_venn import venn3, venn3_circles, venn2
```

### 1 AF2 vs AF3 structures venn diagrams

```python
sns.set_theme()
fig, axes = plt.subplots(1, 2, figsize=(6,3))
v = venn2(ax=axes[0], subsets=(10,0,35), set_labels=("",""), alpha=1, normalize_to=1)
v.get_patch_by_id("10").set_color("tab:red")
v.get_patch_by_id("11").set_color("tab:orange")
v.get_label_by_id("01").set_visible(False)
v.get_label_by_id("10").set_color("white")
axes[0].set_title("AlphaFold 2")
v = venn2(ax=axes[1], subsets=(12,0,12), set_labels=("",""), alpha=1, normalize_to=0.1)
v.get_patch_by_id("10").set_color("gold")
v.get_patch_by_id("11").set_color("tab:green")
v.get_label_by_id("01").set_visible(False)
axes[1].set_title("AlphaFold 3")
plt.show()
```

```python
plt.figure(figsize=(3,3))
v = venn2(subsets=(24,0,12), set_labels=("",""), alpha=1)
v.get_patch_by_id("10").set_color("gold")
v.get_patch_by_id("11").set_color("tab:green")
v.get_label_by_id("01").set_visible(False)
plt.title("AlphaFold 3")
plt.show()
```
